```python
import numpy as np
rng = np.random.default_rng()
prices = np.array([10, 20, 30, 40, 50])

discounted_prices = prices * 0.9
expensive_items = prices[prices >= 35]

print(discounted_prices)
print(expensive_items)
```

```python
print(prices - 10)
```

```python
print(prices[prices >= 25])
```
