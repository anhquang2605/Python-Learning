```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "job": ["Frontend Developer", "AI Engineer", "ML Engineer", "Data Analyst"],
    "salary": [100000, 140000, 150000, 95000],
    "python_required": [False, True, True, True]
})
```

Exercises
- Find jobs paying more than $110,000.
- Find jobs that require Python.
- Sort jobs from highest to lowest salary.
- Create a column that labels jobs as "High" or "Other" salary.
- Count the number of jobs requiring Python.

```python
print(df[df['salary'] > 110000])
```

```python
print(df[df['python_required']])
```

```python
print(df.sort_values('salary', ascending=False))
```

```python
df['pay_rate_type'] = 'Other'
df.loc[df['salary'] > 140000, 'pay_rate_type'] = 'High'
```

```python
df['pay_rate_type'] = np.where(df['salary'] > 140000, 'High', 'Other')
```
