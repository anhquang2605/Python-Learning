```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    'employee_id': range(1, 11),
    'department': [
        'IT', 'IT', 'IT', 'HR', 'HR',
        'HR', 'Sales', 'Sales', 'Sales', 'Sales'
    ],
    'experience': [1, 2, 3, 2, 4, 6, 1, 3, 5, 8],
    'salary': [
        65000, 75000, 85000, 55000, 70000,
        90000, 60000, 80000, 100000, 180000
    ],
    'age': [23, 25, 27, 24, 30, 35, 22, 28, 33, 42]
}

df = pd.DataFrame(data)

df
```

```python
print(df)
head = df.head()
tail = df.tail()
print(head)
print(tail)
df.describe()
```

```python
df['salary'].plot(kind='line')
plt.xlabel("cool")
plt.ylabel('nice')
plt.title("nice chart")
```

```python
df.groupby('department')['salary'].mean().plot(kind='bar')
plt.xlabel('department')
plt.ylabel('salary mean')
plt.title('Comparing average salary between departments')
```

```python
df['salary'].plot(kind='hist', bins=10)
plt.xlabel("salary")
plt.title("salary distribution")
```

```python
df.plot(kind='scatter', x='salary', y='experience')
```
