```python
days = [1, 2, 3, 4, 5]
temperatures = [68, 70, 75, 72, 78]

plt.plot(days, temperatures, marker="o")
plt.title("Daily Temperatures")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.show()
```
