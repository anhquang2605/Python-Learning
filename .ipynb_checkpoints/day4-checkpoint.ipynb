```python
import numpy as np
rng = np.random.default_rng()
scores = np.array([
    [80, 90, 85],
    [70, 75, 80],
    [95, 90, 100]
])

print(scores.mean(axis=0))
print(scores.mean(axis=1))
```

Here, axis=0 calculates a mean for each column, while axis=1 calculates a mean for each row.

Exercises
- Reshape 12 numbers into a 3 × 4 array.
- Calculate the average of each column.
- Normalize an array by subtracting its mean.
- Multiply two compatible matrices using @.

```python
arr1 = rng.integers(12, size=(12,));
```

```python
print(arr1)
```

```python
arr2 = arr1.reshape(3,4)
print(arr2)
```

```python
print(arr2.mean(axis=0))
```

```python
arr2mean = arr2.mean()
```

```python
print(arr2mean)
```

```python
normalizedarr2 = arr2 - arr2mean
```

```python
print(normalizedarr2)
```

```python
arr3 = rng.integers(99, size=(6,8));
print(arr3)
```

```python
arr4 = rng.integers(20, size=(8,3))
print(arr4)
```

```python
print(arr3@arr4)
```
