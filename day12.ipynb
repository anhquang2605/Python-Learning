```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('postings.csv')
print(data)
```

```python
data.isna().sum()
```

```python
data.duplicated(keep=False)
```

```python
data.fillna({'company_name':'Unknown', 'max_salary': data['max_salary'].mean()},inplace=True)
```

```python
pd.to_numeric(data['max_salary'],errors='coerce')
data['pay_period'].fillna('Not Known')
print(data['pay_period'])
```

fillna() wont work because the 'Nan' is literally a string

```python
data['pay_period'] = data['pay_period'].replace('NaN', "Non")
print(data['pay_period'])
```

```python
# Returns a filtered DataFrame containing only matching rows
matching_rows = data[data['title'].str.contains('ML|AI', na=False)]
print(matching_rows.groupby('title').count())
```

```python
thediff = data.groupby('title')['max_salary'].mean().diff()
print(thediff)
```

```python
data[1:10].groupby('title')['max_salary'].mean().plot(kind='bar')
plt.xticks(rotation=45)
```
