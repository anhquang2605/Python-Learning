```python
import pandas as pd

employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4],
    'name': ['Alex', 'Bob', 'Cindy', 'David'],
    'department_id': [101, 102, 101, 103]
})

departments = pd.DataFrame({
    'department_id': [101, 102, 104],
    'department': ['Engineering', 'Marketing', 'Finance']
})
```

```python
merged = pd.merge(employees, departments,
         on='department_id', how='outer')
```

```python
print(merged)
```
